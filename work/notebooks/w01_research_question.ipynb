## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd

# Load the starter dataset
df = pd.read_csv('../../data/raw/content_refresh_anonymized.csv')

# 1. Total number of pages
total_pages = df['content_id'].nunique()

# 2. Number and percentage of declining pages (Lane 2 core focus)
declining_df = df[df['trend_direction'] == 'down']
num_declining = declining_df['content_id'].nunique()
pct_declining = (num_declining / total_pages) * 100

# 3. Total impressions/traffic at risk
total_impressions_at_risk = declining_df['impressions_90d'].sum()

print(f"Total Unique Pages in Dataset: {total_pages:,}")
print(f"Declining Pages (Performance Decay): {num_declining:,} ({pct_declining:.2f}%)")
print(f"Total Impressions at Risk (Last 90 Days): {total_impressions_at_risk:,}")
```
